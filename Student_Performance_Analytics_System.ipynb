```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("students.csv")

print("Student Dataset:")
display(df)
```

```python
print("Dataset shape:")
print(df.shape)
```

```python
print("Column names:")
print(df.columns.tolist())
```

```python
print("Dataset information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
subjects = ["Python", "SQL", "HTML", "CSS", "JavaScript"]

df["Total"] = df[subjects].sum(axis=1)

df["Average"] = df[subjects].mean(axis=1).round(2)

display(df[["Student_ID", "Name", "Total", "Average"]])
```

```python
def calculate_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(calculate_grade)

display(df[["Student_ID", "Name", "Average", "Grade"]])
```

```python
df["Result"] = np.where(
    (df[subjects] >= 40).all(axis=1),
    "Pass",
    "Fail"
)

display(df[["Student_ID", "Name", "Average", "Grade", "Result"]])
```

```python
class_average = df["Average"].mean()

print("Class Average:", round(class_average, 2))
```

```python
topper = df.loc[df["Average"].idxmax()]

print("Top-Performing Student")
print("Name:", topper["Name"])
print("Average:", topper["Average"])
print("Grade:", topper["Grade"])
```

```python
lowest_student = df.loc[df["Average"].idxmin()]

print("Lowest-Performing Student")
print("Name:", lowest_student["Name"])
print("Average:", lowest_student["Average"])
print("Grade:", lowest_student["Grade"])
```

```python
subject_averages = df[subjects].mean().round(2)

print("Average Marks by Subject:")
print(subject_averages)

print("\nHighest-Performing Subject:",
      subject_averages.idxmax())

print("Lowest-Performing Subject:",
      subject_averages.idxmin())
```

```python
result_counts = df["Result"].value_counts()

print("Pass/Fail Summary:")
print(result_counts)
```

```python
print("========== STUDENT PERFORMANCE REPORT ==========")

display(df)

print("\n========== CLASS SUMMARY ==========")
print("Total Students:", len(df))
print("Class Average:", round(df["Average"].mean(), 2))
print("Students Passed:", (df["Result"] == "Pass").sum())
print("Students Failed:", (df["Result"] == "Fail").sum())

print("\n========== TOP PERFORMER ==========")
print(df.loc[df["Average"].idxmax(), ["Name", "Average", "Grade"]])

print("\n========== SUBJECT-WISE AVERAGES ==========")
print(df[subjects].mean().round(2))
```

```python
def calculate_total(marks):
    total = np.sum(marks)
    return total


def calculate_average(marks):
    average = np.mean(marks)
    return round(average, 2)


def calculate_grade(average):
    if average >= 90:
        return "A"
    elif average >= 80:
        return "B"
    elif average >= 70:
        return "C"
    elif average >= 60:
        return "D"
    else:
        return "F"


print("Functions created successfully!")
```

```python
totals = []
averages = []
grades = []

for index, row in df.iterrows():
    marks = row[subjects].to_numpy()

    total = calculate_total(marks)
    average = calculate_average(marks)
    grade = calculate_grade(average)

    totals.append(total)
    averages.append(average)
    grades.append(grade)

df["Total"] = totals
df["Average"] = averages
df["Grade"] = grades

print("Student records processed successfully!")
display(df[["Name", "Total", "Average", "Grade"]])
```

```python
def check_attendance(attendance):
    if attendance >= 75:
        return "Satisfactory"
    else:
        return "Low Attendance"


df["Attendance_Status"] = df["Attendance"].apply(check_attendance)

print("Attendance Analysis:")
display(
    df[["Student_ID", "Name", "Attendance", "Attendance_Status"]]
)
```
